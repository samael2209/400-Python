## Ejercicio 073: Clave con Valor Máximo

### Descripción
Encontrar la clave con el valor máximo en un diccionario de números enteros.

### Conceptos aprendidos
- `.keys()`
- Iteración y comparación

```python
numeros = {1: 'uno', 2: 'dos', 100: 'cien', 1000: 'mil'}
claves = list(numeros.keys())
maximo = claves[0]
for k in claves:
    if k > maximo: maximo = k
print(maximo)
```

## Ejercicio 074: Fusión de Diccionarios

### Descripción
Fusionar dos diccionarios sin usar `update()`.

### Conceptos aprendidos
- Dictionary comprehension
- `set()` para claves únicas

```python
d1 = {1: 'uno', 2: 'dos'}
d2 = {3: 'tres', 4: 'cuatro'}
resultado = {k: d2[k] if k in d2 else d1[k] for k in set(d1) | set(d2)}
print(resultado)
```

## Ejercicio 076: Claves con Valores Iguales

### Descripción
Encontrar claves que tienen el mismo valor en dos diccionarios.

### Conceptos aprendidos
- `keys() &` intersección
- Comparación de valores

```python
d1 = {k: k for k in range(0, 101, 3)}
d2 = {k: k for k in range(0, 101)}
claves_comunes = [k for k in d1.keys() & d2.keys() if d1[k] == d2[k]]
print(claves_comunes)
```

## Ejercicio 077: Suma de Valores en Diccionario Anidado

### Descripción
Calcular la suma de todos los valores en un diccionario anidado.

### Conceptos aprendidos
- Diccionarios anidados
- Doble iteración

## Ejercicio 079: Tupla con Mayor Suma

### Descripción
Encontrar la tupla con la mayor suma de elementos en una lista de tuplas.

### Conceptos aprendidos
- Iteración sobre listas de tuplas
- Función `sum()`

```python
t = [(1, 2), (3, 4), (0, 10), (2, 2)]
tupla_mayor = None
suma_mayor = float("-inf")
for tupla in t:
    s = sum(tupla)
    if s > suma_mayor:
        tupla_mayor = tupla
        suma_mayor = s
print(tupla_mayor)
```

## Ejercicio 080: Combinar Listas de Tuplas

### Descripción
Combinar dos listas de tuplas eliminando duplicados basados en el primer elemento.

### Conceptos aprendidos
- Manipulación manual de listas
- Reserva de espacio

```python
t1 = [(i, i*2) for i in range(3)]
t2 = [(i, i*6) for i in range(9)]
t = [None] * (len(t1) + len(t2))
k = 0
for i in range(len(t1)):
    t[k] = t1[i]
    k += 1
for j in range(len(t2)):
    t[k] = t2[j]
    k += 1
print(t)
```

## Ejercicio 081: Dividir en Pares e Impares

### Descripción
Dividir una lista de números en dos listas de tuplas (pares e impares).

### Conceptos aprendidos
- List comprehension
- `defaultdict`

```python
from collections import defaultdict
lista_numeros = [i for i in range(50)]
resultado = defaultdict(list)
for n in lista_numeros:
    clave = "pares" if n % 2 == 0 else "impares"
    resultado[clave].append((n,))
print("Pares", resultado['pares'])
print("Impares", resultado['impares'])
```

## Ejercicio 082: Tupla Más Común

### Descripción
Encontrar la tupla más común en una lista de tuplas usando un contador.

### Conceptos aprendidos
- Diccionario como contador
- `max()` con `key`

```python
import random
contador = {}
t = [(random.randint(1,5),) for _ in range(20)]
for tupla in t:
    if tupla in contador:
        contador[tupla] += 1
    else:
        contador[tupla] = 1
mayor = max(contador, key=contador.get)
print("Lista:", t)
print("Más común:", mayor)
```

## Ejercicio 083: Intersección de Conjuntos

### Descripción
Calcular la intersección de dos conjuntos sin usar el operador `&`.

### Conceptos aprendidos
- Iteración sobre conjuntos
- Optimización iterando sobre el más pequeño

```python
a = {6, 4, 5, 9}
b = {5, 6, 5, 9}
if len(a) > len(b):
    a, b = b, a
interseccion = set()
for x in a:
    if x in b:
        interseccion.add(x)
print(interseccion)
```

## Ejercicio 084: Verificar Subconjunto

### Descripción
Verificar si un conjunto es subconjunto de otro usando iteración manual.

### Conceptos aprendidos
- Conjuntos congelados (`frozenset`)
- Iteración y banderas booleanas

```python
cp = {
    frozenset({1,2,3}),
    frozenset({2,3,4}),
    frozenset({3,4,5})
}
c1 = {1,2,3}
for s in cp:
    subconjunto = True
    for x in c1:
        if x not in s:
            subconjunto = False
            break
    print(s, subconjunto)
```

## Ejercicio 086: Unión de Múltiples Conjuntos

### Descripción
Calcular la unión de múltiples conjuntos usando una función con `*args`.

### Conceptos aprendidos
- `*args` para número variable de argumentos
- Unión manual de conjuntos

```python
c1 = {1,2,3}
c2 = {9,3,4}
c3 = {4,5,1}
c4 = {1,5,3}
def union(*args):
    r = set()
    for conjunto in args:
        for elemento in conjunto:
            r.add(elemento)
    return r
print(union(c1,c2,c3,c4))
```

## Ejercicio 087: Diferencia Simétrica

### Descripción
Encontrar la diferencia simétrica entre múltiples conjuntos.

### Conceptos aprendidos
- Algoritmo de toggle (agregar/remover)

```python
c1 = {1,2,3}
c2 = {9,3,4}
c3 = {4,5,1}
c4 = {1,5,3}
def diferencia(*args):
    r = set()
    for conjunto in args:
        for elemento in conjunto:
            if elemento in r:
                r.remove(elemento)
            else:
                r.add(elemento)
    return r
print(diferencia(c1,c2,c3,c4))
```

## Ejercicio 088: Eliminar Duplicados Manteniendo Orden

### Descripción
Eliminar elementos duplicados de una lista sin cambiar el orden original.

- Preservación de orden

## Ejercicio 092: Subsecuencia Creciente Más Larga

### Descripción
Encontrar la subsecuencia más larga creciente (LIS) usando programación dinámica y búsqueda binaria.

### Conceptos aprendidos
- Programación dinámica
- `bisect_left` para búsqueda binaria

```python
from bisect import bisect_left
nums = [10,9,2,5,3,7,101,18]
tails = []
for x in nums:
    i = bisect_left(tails, x)
    if i == len(tails):
        tails.append(x)
    else:
        tails[i] = x
print(tails)
print("LIS:", len(tails))
```

Crear un sistema de usuarios donde cada usuario tiene nombre, correo y lista de pedidos.

## Ejercicio 097: CRUD de Productos con Revisiones

### Descripción
Sistema completo CRUD para gestionar productos y sus revisiones de usuarios.

### Conceptos aprendidos
- CRUD completo
- Diccionarios anidados

```python
productos = {
    "producto_1": {"nombre": "Laptop Lenovo", "revisiones": []}
}
def agregar_revision(productos, codigo, usuario, comentario, calificacion):
    if codigo in productos:
        productos[codigo]["revisiones"].append({
            "usuario": usuario, "comentario": comentario, "calificacion": calificacion
        })
def mostrar_productos(productos):
    for codigo, prod in productos.items():
        print(f"{codigo}: {prod['nombre']} - {len(prod['revisiones'])} revisiones")
agregar_revision(productos, "producto_1", "Jordy", "Excelente", 5)
mostrar_productos(productos)
```

## Ejercicio 099: Estudiantes Ordenados por Calificaciones

### Descripción
Ordenar estudiantes según su promedio de calificaciones usando Bubble Sort.

### Conceptos aprendidos
- Bubble Sort
- Cálculo de promedios
- Tuplas de calificaciones

```python
estudiantes = [
    {"nombre": "Jordy", "calificaciones": (9.5, 8.7, 10.0)},
    {"nombre": "Maria", "calificaciones": (9.8, 9.5, 9.7)},
    {"nombre": "Luis", "calificaciones": (8.0, 7.9, 8.4)}
]
promedios = [sum(e["calificaciones"]) / len(e["calificaciones"]) for e in estudiantes]
n = len(estudiantes)
for i in range(n):
    for j in range(n - 1 - i):
        if promedios[j] < promedios[j + 1]:
            estudiantes[j], estudiantes[j + 1] = estudiantes[j + 1], estudiantes[j]
            promedios[j], promedios[j + 1] = promedios[j + 1], promedios[j]
for e, p in zip(estudiantes, promedios):
    print(f"{e['nombre']}: {p:.2f}")
```

## Ejercicio 100: Validación de Correos Electrónicos

### Descripción
Clasificar correos electrónicos en válidos e inválidos.

### Conceptos aprendidos
- Validación básica de strings
- División en listas condicional

```python
correos = [
    "juan@gmail.com", "maria@hotmail.com", "correo_invalido",
    "pedro@", "@gmail.com", "ana123@yahoo.com"
]
validos = []
invalidos = []
for correo in correos:
    if "@" in correo and "." in correo.split("@")[-1]:
        validos.append(correo)
    else:
        invalidos.append(correo)
print("Válidos:", validos)
print("Inválidos:", invalidos)
```

## Ejercicio 101: CRUD de Ciudades con Temperaturas

### Descripción
Administrar un diccionario de ciudades con temperaturas diarias (CRUD completo).

### Conceptos aprendidos
- CRUD completo
- Listas de tuplas como valores

```python
temperaturas_ciudades = {
    "Quito": [("2026-05-10", 22), ("2026-05-11", 24)],
    "Guayaquil": [("2026-05-10", 31), ("2026-05-11", 32)]
}
def agregar_temperatura(diccionario, ciudad, fecha, temperatura):
    if ciudad in diccionario:
        diccionario[ciudad].append((fecha, temperatura))
def mostrar_ciudades(diccionario):
    for ciudad, datos in diccionario.items():
        print(f"{ciudad}:")
        for fecha, temp in datos:
            print(f"  {fecha}: {temp}°C")
agregar_temperatura(temperaturas_ciudades, "Quito", "2026-05-12", 26)
mostrar_ciudades(temperaturas_ciudades)
```

## Ejercicio 102: Elemento Más Común en Lista de Tuplas

### Descripción
Encontrar la tupla más frecuente en una lista y contar sus apariciones.

### Conceptos aprendidos
- Diccionario como contador
- `max()` con `key.get`

```python
datos = [
    ("manzana", "rojo"), ("pera", "verde"), ("manzana", "rojo"),
    ("uva", "morado"), ("pera", "verde"), ("manzana", "rojo")
]
contador = {}
for dato in datos:
    contador[dato] = contador.get(dato, 0) + 1
elemento = max(contador, key=contador.get)
print(f"Más común: {elemento} ({contador[elemento]} veces)")
```

## Ejercicio 103: Gestión de Tareas por Usuario

### Descripción
Gestionar tareas pendientes usando diccionarios con conjuntos como valores (CRUD).

### Conceptos aprendidos
- Diccionarios con conjuntos como valores
- CRUD con estructuras mixtas

```python
tareas_usuarios = {
    "Jordy": {"Estudiar Python", "Hacer ejercicio"},
    "Ana": {"Diseñar interfaz", "Terminar reporte"}
}
def agregar_tarea(diccionario, usuario, tarea):
    if usuario in diccionario:
        diccionario[usuario].add(tarea)
def mostrar_usuarios(diccionario):
    for usuario, tareas in diccionario.items():
        print(f"{usuario}: {', '.join(tareas)}")
agregar_tarea(tareas_usuarios, "Jordy", "Leer sobre APIs")
mostrar_usuarios(tareas_usuarios)
```

## Ejercicio 105: Lista de Invitados con Asistencia

### Descripción
Gestionar invitados a un evento usando diccionarios con conjuntos (CRUD).

### Conceptos aprendidos
- Diferencia de conjuntos
- CRUD con conjuntos

```python
evento = {
    "invitados": {"Jordy", "Ana", "Carlos", "María"},
    "asistieron": set()
}
def registrar_asistencia(evento, nombre):
    if nombre in evento["invitados"]:
        evento["asistieron"].add(nombre)
def mostrar_faltantes(evento):
    faltantes = evento["invitados"] - evento["asistieron"]
    print("Faltantes:", faltantes)
registrar_asistencia(evento, "Jordy")
registrar_asistencia(evento, "Ana")
mostrar_faltantes(evento)
```

## Ejercicio 106: Intersección de Múltiples Conjuntos

### Descripción
Encontrar elementos comunes en múltiples conjuntos almacenados en un diccionario.

### Conceptos aprendidos
- Intersección manual
- Iteración sobre valores del diccionario

```python
conjuntos = {
    "grupo_1": {1, 2, 3, 4, 5},
    "grupo_2": {3, 4, 5, 6, 7},
    "grupo_3": {2, 3, 4, 5, 8}
}
valores = list(conjuntos.values())
resultado = set()
for num in valores[0]:
    if all(num in c for c in valores[1:]):
        resultado.add(num)
print("Intersección:", resultado)
```
