## Ejercicio 073: Clave con Valor Máximo

### Descripción
Encontrar la clave con el valor máximo en un diccionario de números enteros.

### Conceptos aprendidos
- `.keys()`
- Iteración y comparación

```python
numeros = {1: 'uno', 2: 'dos', 100: 'cien', 1000: 'mil'}
claves = list(numeros.keys())
maximo = claves[0]
for k in claves:
    if k > maximo: maximo = k
print(maximo)
```

## Ejercicio 074: Fusión de Diccionarios

### Descripción
Fusionar dos diccionarios sin usar `update()`.

### Conceptos aprendidos
- Dictionary comprehension
- `set()` para claves únicas

```python
d1 = {1: 'uno', 2: 'dos'}
d2 = {3: 'tres', 4: 'cuatro'}
resultado = {k: d2[k] if k in d2 else d1[k] for k in set(d1) | set(d2)}
print(resultado)
```

## Ejercicio 075: Eliminar Claves con Valores Pares

### Descripción
Eliminar todas las claves que tengan valores pares.

### Conceptos aprendidos
- Dictionary comprehension
- Filtrar por valor

```python
numeros = {k: k**2 for k in range(10)}
filtrado = {k: v for k, v in numeros.items() if v % 2 != 0}
print(filtrado)
```

## Ejercicio 076: Claves con Valores Iguales

### Descripción
Encontrar claves que tienen el mismo valor en dos diccionarios.

### Conceptos aprendidos
- `keys() &` intersección
- Comparación de valores

```python
d1 = {k: k for k in range(0, 101, 3)}
d2 = {k: k for k in range(0, 101)}
claves_comunes = [k for k in d1.keys() & d2.keys() if d1[k] == d2[k]]
print(claves_comunes)
```

## Ejercicio 077: Suma de Valores en Diccionario Anidado

### Descripción
Calcular la suma de todos los valores en un diccionario anidado.

### Conceptos aprendidos
- Diccionarios anidados
- Doble iteración

```python
m = {i: {j: i*j for j in range(3, 5)} for i in range(5)}
total = sum(valor for subdict in m.values() for valor in subdict.values())
print(total)
```

## Ejercicio 078: Ordenar Lista de Tuplas

### Descripción
Ordenar una lista de tuplas en función del segundo elemento.

### Conceptos aprendidos
- Algoritmo QuickSort
- Ordenamiento personalizado

```python
tuplas = [(1, 5), (2, 3), (4, 1), (0, 9)]
tuplas.sort(key=lambda x: x[1])
print(tuplas)
```
