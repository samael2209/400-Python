## Ejercicio 073: Clave con Valor Máximo

### Descripción
Encontrar la clave con el valor máximo en un diccionario de números enteros.

### Conceptos aprendidos
- `.keys()`
- Iteración y comparación

```python
numeros = {1: 'uno', 2: 'dos', 100: 'cien', 1000: 'mil'}
claves = list(numeros.keys())
maximo = claves[0]
for k in claves:
    if k > maximo: maximo = k
print(maximo)
```

## Ejercicio 074: Fusión de Diccionarios

### Descripción
Fusionar dos diccionarios sin usar `update()`.

### Conceptos aprendidos
- Dictionary comprehension
- `set()` para claves únicas

```python
d1 = {1: 'uno', 2: 'dos'}
d2 = {3: 'tres', 4: 'cuatro'}
resultado = {k: d2[k] if k in d2 else d1[k] for k in set(d1) | set(d2)}
print(resultado)
```

## Ejercicio 075: Eliminar Claves con Valores Pares

### Descripción
Eliminar todas las claves que tengan valores pares.

### Conceptos aprendidos
- Dictionary comprehension
- Filtrar por valor

```python
numeros = {k: k**2 for k in range(10)}
filtrado = {k: v for k, v in numeros.items() if v % 2 != 0}
print(filtrado)
```

## Ejercicio 076: Claves con Valores Iguales

### Descripción
Encontrar claves que tienen el mismo valor en dos diccionarios.

### Conceptos aprendidos
- `keys() &` intersección
- Comparación de valores

```python
d1 = {k: k for k in range(0, 101, 3)}
d2 = {k: k for k in range(0, 101)}
claves_comunes = [k for k in d1.keys() & d2.keys() if d1[k] == d2[k]]
print(claves_comunes)
```

## Ejercicio 077: Suma de Valores en Diccionario Anidado

### Descripción
Calcular la suma de todos los valores en un diccionario anidado.

### Conceptos aprendidos
- Diccionarios anidados
- Doble iteración

```python
m = {i: {j: i*j for j in range(3, 5)} for i in range(5)}
total = sum(valor for subdict in m.values() for valor in subdict.values())
print(total)
```

## Ejercicio 079: Tupla con Mayor Suma

### Descripción
Encontrar la tupla con la mayor suma de elementos en una lista de tuplas.

### Conceptos aprendidos
- Iteración sobre listas de tuplas
- Función `sum()`

```python
t = [(1, 2), (3, 4), (0, 10), (2, 2)]
tupla_mayor = None
suma_mayor = float("-inf")
for tupla in t:
    s = sum(tupla)
    if s > suma_mayor:
        tupla_mayor = tupla
        suma_mayor = s
print(tupla_mayor)
```

## Ejercicio 080: Combinar Listas de Tuplas

### Descripción
Combinar dos listas de tuplas eliminando duplicados basados en el primer elemento.

### Conceptos aprendidos
- Manipulación manual de listas
- Reserva de espacio

```python
t1 = [(i, i*2) for i in range(3)]
t2 = [(i, i*6) for i in range(9)]
t = [None] * (len(t1) + len(t2))
k = 0
for i in range(len(t1)):
    t[k] = t1[i]
    k += 1
for j in range(len(t2)):
    t[k] = t2[j]
    k += 1
print(t)
```

## Ejercicio 081: Dividir en Pares e Impares

### Descripción
Dividir una lista de números en dos listas de tuplas (pares e impares).

### Conceptos aprendidos
- List comprehension
- `defaultdict`

```python
from collections import defaultdict
lista_numeros = [i for i in range(50)]
resultado = defaultdict(list)
for n in lista_numeros:
    clave = "pares" if n % 2 == 0 else "impares"
    resultado[clave].append((n,))
print("Pares", resultado['pares'])
print("Impares", resultado['impares'])
```

## Ejercicio 082: Tupla Más Común

### Descripción
Encontrar la tupla más común en una lista de tuplas usando un contador.

### Conceptos aprendidos
- Diccionario como contador
- `max()` con `key`

```python
import random
contador = {}
t = [(random.randint(1,5),) for _ in range(20)]
for tupla in t:
    if tupla in contador:
        contador[tupla] += 1
    else:
        contador[tupla] = 1
mayor = max(contador, key=contador.get)
print("Lista:", t)
print("Más común:", mayor)
```

## Ejercicio 083: Intersección de Conjuntos

### Descripción
Calcular la intersección de dos conjuntos sin usar el operador `&`.

### Conceptos aprendidos
- Iteración sobre conjuntos
- Optimización iterando sobre el más pequeño

```python
a = {6, 4, 5, 9}
b = {5, 6, 5, 9}
if len(a) > len(b):
    a, b = b, a
interseccion = set()
for x in a:
    if x in b:
        interseccion.add(x)
print(interseccion)
```

## Ejercicio 084: Verificar Subconjunto

### Descripción
Verificar si un conjunto es subconjunto de otro usando iteración manual.

### Conceptos aprendidos
- Conjuntos congelados (`frozenset`)
- Iteración y banderas booleanas

```python
cp = {
    frozenset({1,2,3}),
    frozenset({2,3,4}),
    frozenset({3,4,5})
}
c1 = {1,2,3}
for s in cp:
    subconjunto = True
    for x in c1:
        if x not in s:
            subconjunto = False
            break
    print(s, subconjunto)
```

## Ejercicio 086: Unión de Múltiples Conjuntos

### Descripción
Calcular la unión de múltiples conjuntos usando una función con `*args`.

### Conceptos aprendidos
- `*args` para número variable de argumentos
- Unión manual de conjuntos

```python
c1 = {1,2,3}
c2 = {9,3,4}
c3 = {4,5,1}
c4 = {1,5,3}
def union(*args):
    r = set()
    for conjunto in args:
        for elemento in conjunto:
            r.add(elemento)
    return r
print(union(c1,c2,c3,c4))
```

## Ejercicio 087: Diferencia Simétrica

### Descripción
Encontrar la diferencia simétrica entre múltiples conjuntos.

### Conceptos aprendidos
- Algoritmo de toggle (agregar/remover)

```python
c1 = {1,2,3}
c2 = {9,3,4}
c3 = {4,5,1}
c4 = {1,5,3}
def diferencia(*args):
    r = set()
    for conjunto in args:
        for elemento in conjunto:
            if elemento in r:
                r.remove(elemento)
            else:
                r.add(elemento)
    return r
print(diferencia(c1,c2,c3,c4))
```

## Ejercicio 088: Eliminar Duplicados Manteniendo Orden

### Descripción
Eliminar elementos duplicados de una lista sin cambiar el orden original.

### Conceptos aprendidos
- Set para seguimiento de vistos
- Preservación de orden

```python
import random
l1 = [random.randint(1,7) for _ in range(30)]
r = []
vistos = set()
for x in l1:
    if x not in vistos:
        r.append(x)
        vistos.add(x)
print("Original:", l1)
print("Sin duplicados:", r)
```

## Ejercicio 089: Elemento Más Repetido

### Descripción
Encontrar el elemento que más se repite en una lista usando un contador.

### Conceptos aprendidos
- Diccionario como frecuencia
- `max()` con `key.get`

```python
import random
numeros = [random.randint(1,5) for _ in range(200)]
contador = {}
for n in numeros:
    if n in contador:
        contador[n] += 1
    else:
        contador[n] = 1
mayor = max(contador, key=contador.get)
print("Más común:", mayor)
```

## Ejercicio 091: Rotación a la Izquierda

### Descripción
Realizar una rotación a la izquierda en una lista (desplazar elementos hacia la izquierda).

### Conceptos aprendidos
- Desplazamiento manual de elementos
- Rotación in-place

```python
l = [1,2,3,4,5]
primero = l[0]
for i in range(len(l) - 1):
    l[i] = l[i+1]
l[-1] = primero
print(l)
```

## Ejercicio 092: Subsecuencia Creciente Más Larga

### Descripción
Encontrar la subsecuencia más larga creciente (LIS) usando programación dinámica y búsqueda binaria.

### Conceptos aprendidos
- Programación dinámica
- `bisect_left` para búsqueda binaria

```python
from bisect import bisect_left
nums = [10,9,2,5,3,7,101,18]
tails = []
for x in nums:
    i = bisect_left(tails, x)
    if i == len(tails):
        tails.append(x)
    else:
        tails[i] = x
print(tails)
print("LIS:", len(tails))
```
